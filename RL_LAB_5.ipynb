```python
import gymnasium as gym
import random


def run_random_agent(env_name):
    print("\n========================================")
    print(f"Environment: {env_name}")
    print("========================================")

    # Step 1: Create environment
    env = gym.make(env_name)

    # Step 2: Reset environment
    state, info = env.reset()

    print("Initial State:", state)
    print("Action Space:", env.action_space)

    total_reward = 0
    step_count = 0

    # Step 3-8: Agent-environment interaction
    while True:

        # Step 5: Select a random action
        action = env.action_space.sample()

        # Step 6: Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        step_count += 1
        total_reward += reward

        print(
            f"Step {step_count}: "
            f"State = {state}, "
            f"Action = {action}, "
            f"Reward = {reward}, "
            f"Next State = {next_state}"
        )

        # Update state
        state = next_state

        # Step 8: Check whether episode is finished
        if terminated or truncated:
            break

    # Step 9: Calculate average reward
    average_reward = total_reward / step_count

    print("\nEpisode Finished")
    print("Total Reward:", total_reward)
    print("Number of Steps:", step_count)
    print("Average Reward:", round(average_reward, 4))

    env.close()


# Run Random Agent on all three environments
run_random_agent("CartPole-v1")
run_random_agent("MountainCar-v0")
run_random_agent("FrozenLake-v1")
```
