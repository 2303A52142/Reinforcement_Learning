```python
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B - Initialize N, R, Q and UCB
N = {machine: 0 for machine in machines}      # Participation count
R = {machine: 0 for machine in machines}      # Total reward
Q = {machine: 0.0 for machine in machines}    # Estimated Q-value
UCB = {machine: 0.0 for machine in machines}  # UCB value

total_reward = 0

print("----- UCB Multi-Armed Bandit -----")

for play in range(1, N_play + 1):

    # Part C - Calculate UCB
    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine with highest UCB
    selected_machine = max(machines, key=lambda x: UCB[x])

    print(f"\nPlay {play}")
    print(f"UCB Values: A = {UCB['A']:.4f}, "
          f"B = {UCB['B']:.4f}, "
          f"C = {UCB['C']:.4f}")

    print(f"Selected Machine = {selected_machine}")

    # Part D - Reward Collection
    reward = float(input(f"Enter reward for machine {selected_machine}: "))

    # Part E - Update N, R and Q
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"N({selected_machine}) = {N[selected_machine]}")
    print(f"R({selected_machine}) = {R[selected_machine]:.2f}")
    print(f"Q({selected_machine}) = {Q[selected_machine]:.4f}")


# Part F - Display final results
print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t{R[machine]:.2f}\t"
          f"{UCB[machine]:.4f}\t\t{Q[machine]:.4f}")

print(f"\nTotal Reward = {total_reward:.2f}")
```
