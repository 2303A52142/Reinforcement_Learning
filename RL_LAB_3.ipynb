```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and plays
M = 3
N_play = 10
epsilon = 0.2

# Part B – Initialize N, R and Q tables
N = {machine: 0 for machine in machines}   # Participation count
R = {machine: [] for machine in machines}  # Rewards received
Q = {machine: 0.0 for machine in machines}  # Estimated reward

# Fixed seed for reproducible results
random.seed(10)

total_reward = 0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("-------------------------------------------")

for play in range(1, N_play + 1):

    # Part C – Modified ε-Greedy Decision
    r = random.random()

    # Find machine with highest Q-value
    highest_q = max(Q.values())

    # Machines having the highest Q-value
    highest_machines = [
        machine for machine in machines
        if Q[machine] == highest_q
    ]

    # Exploration
    if r < epsilon:

        # Select a machine except the highest estimated machine
        candidates = [
            machine for machine in machines
            if machine not in highest_machines
        ]

        # If all machines have the same Q-value initially,
        # randomly choose one machine
        if not candidates:
            selected_machine = random.choice(machines)
        else:
            selected_machine = random.choice(candidates)

        decision = "Exploration"

    # Exploitation
    else:
        selected_machine = random.choice(highest_machines)
        decision = "Exploitation"

    # Part D – Reward Collection
    print(f"\nPlay {play}")
    print(f"Random number r = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Selected machine = {selected_machine}")

    reward = float(input(f"Enter reward for machine {selected_machine}: "))

    # Part E – Update N, R and Q
    N[selected_machine] += 1
    R[selected_machine].append(reward)

    # Incremental sample-average formula:
    # Q_new = Q_old + (Reward - Q_old) / N
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"Updated Q({selected_machine}) = "
          f"{Q[selected_machine]:.2f}")


# Part F – Display Results
print("\n===========================================")
print("FINAL RESULTS")
print("===========================================")

print("\nMachine\tN (Count)\tRewards\t\tQ-value")

for machine in machines:
    print(
        f"{machine}\t{N[machine]}\t\t"
        f"{R[machine]}\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)
print("Average Reward =", total_reward / N_play)
```
